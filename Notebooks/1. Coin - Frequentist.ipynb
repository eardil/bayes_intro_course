## Coin example

### Confidence intervals

```
ax = plt.axes()

plt.bar(
    range(11),
    binom.pmf(range(11), n=10, p=1/2))
ax.set_xticklabels([str(x/10) for x in range(11)])

plt.xlabel('Proportion of heads')
plt.ylabel('Probability')

plt.title('n=10')

plt.vlines(binom.ppf(0.975, n=10, p=1/2), ymin=0, ymax=0.2, linestyle='--', color='r')
plt.vlines(binom.ppf(0.025, n=10, p=1/2), ymin=0, ymax=0.2, linestyle='--', color='r')

plt.show()
```

**Simulating confidence intervals**

```
binom.rvs(n=10, p=1/2, size=1)
```

Let's run 100 experiments and see what the confidence intervals look like

```
# np.random.seed(42)
ax = plt.axes()

plt.hist(binom.rvs(n=10, p=1/2, size=100), bins=[x+0.5 for x in range(11)])

ax.set_xticklabels([str(x/10) for x in range(11)])

plt.xlabel('Number of heads')
plt.ylabel('Probability')

plt.vlines(binom.ppf(0.975, n=10, p=1/2), ymin=0, ymax=20, linestyle='--', color='r')
plt.vlines(binom.ppf(0.025, n=10, p=1/2), ymin=0, ymax=20, linestyle='--', color='r')

plt.xlim(0,10)

plt.show()
```

More data = more power

```
ax = plt.axes()

plt.bar(
    range(1001),
    binom.pmf(range(1001), n=1000, p=1/2))

ax.set_xticklabels([str(x/10) for x in range(1001)])

plt.xlabel('Proportion of heads')
plt.ylabel('Probability')

plt.title('n=1000')

plt.vlines(binom.ppf(0.975, n=1000, p=1/2), ymin=0, ymax=0.02, linestyle='--', color='r')
plt.vlines(binom.ppf(0.025, n=1000, p=1/2), ymin=0, ymax=0.02, linestyle='--', color='r')

plt.show()
```

```
from scipy.stats import norm

ax = plt.axes()

plt.bar(
    range(11),
    binom.pmf(range(11), n=10, p=1/2))
ax.set_xticklabels([str(x/10) for x in range(11)])

plt.xlabel('Proportion of heads')
plt.ylabel('Probability')

plt.title('n=10')

plt.vlines(binom.ppf(0.975, n=10, p=1/2), ymin=0, ymax=0.2, linestyle='--', color='r')
plt.vlines(binom.ppf(0.025, n=10, p=1/2), ymin=0, ymax=0.2, linestyle='--', color='r')

plt.plot(
    np.linspace(0,10,1000),
    norm.pdf(np.linspace(0,10,1000), loc=10*(1/2), scale=np.sqrt(10*(1/2)*(1-1/2))), 'g')

plt.show()
```
