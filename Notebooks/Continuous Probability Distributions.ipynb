```python
from scipy.stats import norm
import matplotlib.pyplot as plt
import numpy as np
```

## Cumulative probability function vs density

```python
x = np.linspace(-5,5,1000)
y = norm.cdf(x)

plt.plot(x,y)
plt.title('Cumulative probability')
plt.xlabel('X')
plt.ylabel('Probability')
plt.show()
```

```python
x = np.linspace(-5,5,1000)
y = norm.cdf(x)

x0 = 0.4

plt.plot(x,y)
plt.title('Cumulative probability')
plt.xlabel('X')
plt.ylabel('Probability')
plt.grid(alpha=0.3)
plt.hlines(y=norm.cdf(x0), xmin=-5, xmax=x0, linestyle='--')
plt.vlines(x=x0, ymin=0, ymax=norm.cdf(x0),linestyle='--')
plt.annotate('prob={:.2}'.format(norm.cdf(x0)), (-5, norm.cdf(x0)+0.02))
plt.annotate(f'x={x0}', (x0+0.2, 0.02))
plt.show()
```

```python
x = np.linspace(-5,5,1000)
y = norm.pdf(x)

x0 = 1

plt.plot(x,y)
plt.title('Cumulative probability')
plt.xlabel('X')
plt.ylabel('Probability')
plt.grid(alpha=0.3)
plt.fill_between(x[x<x0], norm.pdf(x[x<x0]), color='red', alpha=0.4)
plt.annotate('{:.2}'.format(norm.cdf(x0)),(-0.5,0.2))
plt.show()
```

## Proportionality

Normal distribution:

$$ f(x|\mu, \sigma) = \frac{1}{\sqrt{2\pi}\sigma} e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}$$

or

$$ f(x|\mu, \sigma) \propto e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}$$

```python
mu_ex = 0
sigma_ex = 1
```

We can't talk about probabilities in this case:

```python
from scipy import integrate

def prop_norm(x, mu=mu_ex, sigma=sigma_ex):
  return np.exp(-1/2*(((x-mu)/sigma)**2))

def integ_prop(x, mu=mu_ex, sigma=sigma_ex):
  return integrate.quad(lambda y: prop_norm(y, mu, sigma), -100, x)[0]

x = np.linspace(-5,5,1000)
y = prop_norm(x)

x0 = 1

plt.plot(x,[integ_prop(i) for i in x])
plt.grid(alpha=0.3)
plt.ylabel('?')
plt.hlines(y=integ_prop(x0), xmin=-5, xmax=x0, linestyle='--')
plt.vlines(x=x0, ymin=0, ymax=integ_prop(x0),linestyle='--')
plt.annotate('?={:.2}'.format(integ_prop(x0)), (-5, integ_prop(x0)+0.1))
plt.annotate(f'x={x0}', (x0+0.2, 0.02))
plt.show()

plt.plot(x,y)
plt.title('Normal density vs Proportional function')
plt.xlabel('X')
plt.ylabel('?')
plt.grid(alpha=0.3)
plt.fill_between(x[x<x0], prop_norm(x[x<x0]), color='red', alpha=0.4)
plt.annotate('{:.2}'.format(integ_prop(x0)),(-0.5,0.2))
plt.show()
```

We need the cumulative probability to reach exactly 1

```python
integ_prop(np.inf)
```

```python
from scipy import integrate

def prop_norm(x, mu=mu_ex, sigma=sigma_ex):
  return np.exp(-1/2*(((x-mu)/sigma)**2))

def integ_prop(x, mu=mu_ex, sigma=sigma_ex):
  return integrate.quad(lambda y: prop_norm(y, mu, sigma), -np.inf, x)[0]

constant = integ_prop(np.inf)

def dist_norm(x, mu=mu_ex, sigma=sigma_ex):
  return prop_norm(x, mu, sigma) / constant

def integ_norm(x, mu=mu_ex, sigma=sigma_ex):
  return integrate.quad(lambda y: dist_norm(y, mu, sigma), -np.inf, x)[0]

x = np.linspace(-5,5,1000)
y = dist_norm(x)

x0 = 1

plt.plot(x,[integ_norm(i) for i in x])
plt.grid(alpha=0.3)
plt.ylabel('?')
plt.hlines(y=integ_norm(x0), xmin=-5, xmax=x0, linestyle='--')
plt.vlines(x=x0, ymin=0, ymax=integ_norm(x0),linestyle='--')
plt.annotate('?={:.2}'.format(integ_norm(x0)), (-5, integ_norm(x0)+0.1))
plt.annotate(f'x={x0}', (x0+0.2, 0.02))
plt.show()

plt.plot(x,y)
plt.title('Normal density vs Proportional function')
plt.xlabel('X')
plt.ylabel('?')
plt.grid(alpha=0.3)
plt.fill_between(x[x<x0], dist_norm(x[x<x0]), color='red', alpha=0.4)
plt.annotate('{:.2}'.format(integ_norm(x0)),(-0.5,0.2))
plt.show()
```
